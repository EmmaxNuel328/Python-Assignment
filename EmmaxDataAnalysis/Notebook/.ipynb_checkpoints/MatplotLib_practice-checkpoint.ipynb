```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
x = [5,2,9,4,7]
y = [10,5,8,4,2]
```

```python
plt.plot(x,y)
plt.show()
```

```python
first_random = np.random.uniform(0,1,100)
second_random = np.random.uniform(0,1,100)
plt.scatter(first_random,second_random, color = 'blue')
```

```python
boys = np.array([80,85,90,95,100,105,110,115,120,125])
girls = np.array([240,250,260,270,280,290,300,310,320,330])
```

```python
plt.plot(boys,girls)
plt.title("age",fontsize=14)
plt.xlabel("boys", fontsize=12)
plt.ylabel("girls",fontsize=12)
plt.grid()
plt.show()
```

```python
plt.subplot(1,2,1)
plt.plot(boys,girls)
plt.title("BOYS")
plt.subplot(1,2,2)
plt.plot(boys,girls)
plt.title("GIRLS")
plt.suptitle("Store Data")
plt.tight_layout()
plt.show()
```

```python
alpha = np.array(["A","B","C","D"])
numeric = np.array([3,8,1,10])
plt.bar(alpha,numeric)
plt.show()
```

```python
random_numbers = np.random.normal(170,10,250)
plt.hist(random_numbers)
plt.show()
```

```python
pie_numbers = np.array([35,25,25,15])

plt.pie(pie_numbers)
plt.show()
```

```python
pie_numbers = np.array([35,25,25,15,40,28,30])
labels = ["Apples","Bananas","Cherries","Dates","Pineapple","Grape","Guava"]
explodes = [0,0,0,0,0.2,0,0]
plt.pie(pie_numbers, labels=labels, explode=explodes)
plt.legend(title="Fruits")
plt.show()
```

```python
pie_numbers = np.array([35,25,25,15,40,28,30])
labels = ["Apples","Bananas","Cherries","Dates","Pineapple","Grape","Guava"]

plt.bar(labels,pie_numbers)
# plt.grid()
plt.show()
```

```python
dataset_1 = np.random.normal(100,10,200)
dataset_2 = np.random.normal(80,30,200)
dataset_3 = np.random.normal(90,20,200)
data_to_plot = [dataset_1,dataset_2,dataset_3]
plt.boxplot(data_to_plot)
plt.xticks([1,2,3],['Group A','Group B', 'Group C'])
plt.title("Comparison of multiple Groups")
plt.xlabel("Category")
plt.ylabel("Measurement")
plt.grid()
plt.show()
```
