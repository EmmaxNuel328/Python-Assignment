1. Retrieve the first 10 rows from the products table to understand the available columns.

```python
conn = duckdb.connect()
conn.sql("""
        SELECT * FROM products
        LIMIT 10;
    """)
```

2. Find all unique department names from the departments table.

```python
conn.sql("""
        SELECT DISTINCT department FROM departments
        
    """)
```

3. Count the total number of orders placed in the orders dataset.

```python
conn.sql("""
        SELECT count(*) FROM orders;
        
    """)
```

4. List all product names found specifically in Aisle ID 61

```python
conn.sql("""
        SELECT product_name FROM products Where aisle_id == 61;
        
    """)
```

5. Display all orders sorted by the hour of the day they were placed (highest to lowest).

```python
conn.sql("""
        SELECT * FROM orders
        ORDER BY order_hour_of_day DESC;
    """)
```
